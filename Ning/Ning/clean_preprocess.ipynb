```python
# 2 Handling Contractions
import contractions

def fix_contractions(data):
    expanded_corpus = [contractions.fix(text) for text in data]
    return expanded_corpus
```

# **3. Standardizing accent characters**

```python
import unicodedata

def standardize_accented_chars(text):
    return unicodedata.normalize('NFKD', text).encode('ascii', 'ignore').decode('utf-8', 'ignore')
```

```python
input_text = "Somë words such as résumé, café, protest, divorcé, coördinate, exposé, latté."
print(standardize_accented_chars(input_text))
```

# **4. Noise Removal**

```python
# 1 Removing html tags

def remove_html_tagst_regex(text):
    """Remove HTML tags using regular expression"""
    clean = re.compile('<.*?>')
    return re.sub(clean, '', text)
```

```python
html_text = "<p>Hello, <b>world</b>!</p> <a href='#'>Click here</a>"
clean_text = remove_html_tagst_regex(html_text)
print(clean_text)
```

```python
# 3. Convert the numbers into words

import inflect

p = inflect.engine()

#convert numbers into words
def convert_number(text):
    #split string into list of words
    temp_str = text.split()
    #initialise empty list
    new_string = []

    for word in temp_str:
        # if word is a digit, convert the digit
        # to numbers and append into the new_string list
        if word.isdigit():
            temp=p.number_to_words(word)
            new_string.append(temp)
        # append the word as it is
        else:
            new_string.append(word)

    # join the words of the new_string to form a string
    temp_str = ' '.join(new_string)
    return temp_str


input_text = 'There are 3 balls in this bag, and 12 in the other one.'
convert_number(input_text)
```

```python
# 4. Removing punctuation et special characters

regular_punct = list(string.punctuation)
print(regular_punct)
```

```python
def remove_punctuation(text, punct_list):
    for punc in punct_list:
        if punc in text:
            text = text.replace(punc, '')
    return text.strip()

remove_punctuation("Hey @Lisa, did you know that the summer break is coming ? Amazing right !! It's only 5 more days !!", regular_punct)
```

```python
# 6. Handling emoji/emoticon ---> for sentiment analysis, emoji/emoticon is usually not removed

def remove_emojis_regex(text):
    """Remove emojis using Unicode emoji ranges."""
    #Unicode ranges for emojis
    emoji_pattern = re.compile("["
        u"\U0001F600-\U0001F64F" #emotion
        u"\U0001F3006\U0001F5FF"
        u"\U0001F680-\U0001F6FF"
        u"\U0001F1E0-\U0001F1FF"
        u"\U00002702-\U000027B2"
        u"\U000024C2-\U0001F251"
        u"\U0001f926-\U0001f937"
        u"\U00010000-\U0010ffff"
        "]+", flags=re.UNICODE)
    
    return emoji_pattern.sub(r'', text)
```

```python
# Example
text = "Hello! ☺️ How are you? 🎉 I'm in NYC 🏴‍☠️"
clean_text = remove_emojis_regex(text)
print(clean_text)
```

```python
# 7. Removal of URLs or other irrelevant hints

#Define a regex pattern to match URLs 
url_pattern = re.compile(r'https?://\S+')

#Define a function to remove URLs from text
def remove_urls(text):
    return url_pattern.sub('', text) #replace the url with word "this link"

text = "Visit https://www.google.com for more information."

remove_urls(text)
```

```python
url_pattern = []
url_pattern.append(re.compile(r'https:?//\S+'))
url_pattern.append(re.compile(r'http:?//\S+'))
url_pattern.append(re.compile(r'www.\S+'))

def remove_urls2(text):
    for pattern in url_pattern:
        text=pattern.sub('<this link>', text)
    return text


input_text = "Visit the following link : www.google.com"
print(remove_urls2(input_text))
```

```python
# The required librairies

import nltk 
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

# Download the necessary NLTK datasets
nltk.download('punkt')
nltk.download('stopwords')

#remove stopwords function
def remove_stopwords(text):
    stop_words = set(stopwords.words("english"))
    word_tokens = word_tokenize(text)

    #Filter out the stopwords
    filtered_text = [word for word in word_tokens if word.lower() not in stop_words]

    return filtered_text


example_text = "This is a sample sentence and we are going to remove the stopwords from this"
print("Before stopwords : ", example_text)
print("After removing stopwords : ", remove_stopwords(example_text))
```

# **7. Stemming and Lemmatization**

```python
from nltk.stem.porter import PorterStemmer
from nltk.tokenize import word_tokenize
```

```python
# Stemming

from nltk.stem.porter import PorterStemmer
from nltk.tokenize import word_tokenize
stemmer = PorterStemmer()

#stem words in the list of tokenized words
def stem_words(text):
    word_tokens = word_tokenize(text)
    stems = [stemmer.stem(word) for word in word_tokens]
    return stems

text = "data science uses scientific methods algorihms and many types of processes"
stem_words(text)
```

```python
# Lemmatization using NLTK

import nltk
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize


#Download the necessary NLTK datasets
nltk.download('punkt')
nltk.download('wordnet')

#Initialize the WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

#Function to lemmatize words in the list of tokenized words
def lemma_words(text):
    word_tokens = word_tokenize(text)
    lemmas = [lemmatizer.lemmatize(word) for word in word_tokens]
    return lemmas


#Example text

text = "The quick brown foxes are jumping over the lazy dogs."
print(lemma_words(text))
```
