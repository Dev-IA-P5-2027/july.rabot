```python
import nltk
import string
import re
import pandas as pd
import contractions
import unicodedata
from autocorrect import Speller
import inflect
import emoji
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem.porter import PorterStemmer
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
```

```python
# lowercasing
def text_lowercase(texte):
    return texte.lower()

df['Message_body'] = df['Message_body'].apply(text_lowercase)
```

```python
df.head()
```

```python
def fix_contractions(texte):
    return contractions.fix(str(texte))


df['Message_body'] = df['Message_body'].apply(fix_contractions)
```

```python
df.head()
```

```python
def correct_with_autocorrect(text, lang='en'):
    """Correct spelling using autocorrect library for English."""
    spell = Speller(lang=lang)
    return spell(text)

df['Message_body'] = df['Message_body'].apply(correct_with_autocorrect)
```

```python
df.head()
```

```python
def standardize_accented_chars(text):
    return unicodedata.normalize('NFKD', text).encode('ascii', 'ignore').decode('utf-8', 'ignore')

df['Message_body'] = df['Message_body'].apply(standardize_accented_chars)
```

```python
df.head()
```

```python
def remove_html_tagst_regex(text):
    """Remove HTML tags using regular expression"""
    clean = re.compile('<.*?>')
    return re.sub(clean, '', text)


df['Message_body'] = df['Message_body'].apply(remove_html_tagst_regex)
```

```python
df.head()
```

```python
def remove_numbers(text):
    result = re.sub(r'\d+', '<number>', text)
    return result

df['Message_body'] = df['Message_body'].apply(remove_numbers)
```

```python
def remove_numbers(text):
    result = re.sub(r'\d{7,}', '<number>', text)
    return result

df['Message_body'] = df['Message_body'].apply(remove_numbers)
```

```python
p = inflect.engine()

#convert numbers into words
def convert_number(text):
    #split string into list of words
    temp_str = text.split()
    #initialise empty list
    new_string = []

    for word in temp_str:
        # if word is a digit, convert the digit
        # to numbers and append into the new_string list
        if word.isdigit():
            temp=p.number_to_words(word)
            new_string.append(temp)
        # append the word as it is
        else:
            new_string.append(word)

    # join the words of the new_string to form a string
    temp_str = ' '.join(new_string)
    return temp_str

df['Message_body'] = df['Message_body'].apply(convert_number)
```

```python
df.head()
```

```python
regular_punct = list(string.punctuation)
print(regular_punct)
```

```python
def remove_punctuation(text, punct_list):
    text = str(text)
    for punc in punct_list:
        if punc in text:
            text = text.replace(punc, '')
    return text.strip()


df['Message_body'] = df['Message_body'].apply(remove_punctuation, args=(regular_punct,))
```

```python
df.head()
```

```python
def remove_whitespace(text):
        return " ".join(text.split())

df['Message_body'] = df['Message_body'].apply(remove_whitespace)
```

```python
def remove_emojis_emoji_library(text):
    """Remove emojis using the emoji library."""
    return emoji.replace_emoji(text, replace = '')

df['Message_body'] = df['Message_body'].apply(remove_emojis_emoji_library)
```

```python
url_pattern = []
url_pattern.append(re.compile(r'https:?//\S+'))
url_pattern.append(re.compile(r'http:?//\S+'))
url_pattern.append(re.compile(r'www.\S+'))

def remove_urls2(text):
    for pattern in url_pattern:
        text=pattern.sub('<this link>', text)
    return text

df['Message_body'] = df['Message_body'].apply(remove_urls2)
```

```python

nltk.download('punkt') 
nltk.download('punkt_tab')


df['Message_body'] = df['Message_body'].apply(nltk.word_tokenize)
```

```python
stop_words = set(stopwords.words("english"))


def remove_stopwords(text):
    if type(text) == list:
        filtered_text = [word for word in text if word.lower() not in stop_words]
        return filtered_text
    
    return text


df['Message_body'] = df['Message_body'].apply(remove_stopwords)
```

```python
stemmer = PorterStemmer()

def stem_words(text):
    if type(text) == list:
        stems = [stemmer.stem(word) for word in text]
        return stems

    return text

df['Message_body'] = df['Message_body'].apply(stem_words)

df.head()
```

```python
nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

def lemma_words(liste_de_mots):
    if type(liste_de_mots) == list:
        lemmas = [lemmatizer.lemmatize(word) for word in liste_de_mots]
        return lemmas
    
    return liste_de_mots

df['Message_body'] = df['Message_body'].apply(lemma_words)

df.head()
```
