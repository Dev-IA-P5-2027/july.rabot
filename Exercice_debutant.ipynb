**Écrire dans une cellule un programme qui demande à l'utilisateur de rentrer un rayon et la hauter, puis , qui calcule le volume d'un cône droit. (Rappel : Vcone = 1/3 * pi * r^2 * h) Bis : mettre ce programme dans une fonction volume_cone() et tester cette fonction**

```python
from math import pi

def volume_cone(rayon, hauteur):
    try:
        Volume = 1/3 * pi * rayon*rayon * hauteur
        return Volume
    except ValueError:
        return "Recommences tes valeurs sont pourris"

r = int(input("Entrez la valeur du rayon"))
h = int(input("Entrez la valeur de la hauteur"))
resultat_volume = volume_cone(r,h)
print(resultat_volume)
```

**Écrire une fonction qui convertit en mètres par seconde et en km/h une vitesse fournie en paramètres par l'utilisateur en miles/heure (1 mile = 1609 mètres). Ce résultat sera retourné par la fonction.**

```python
vitesse = int(input("Entrez votre vitesse en miles par heure : "))
print(f"Vous voulez convertir {vitesse} miles/heure")


def convertisseur_ms(mph):
    MS = mph * 0.44704
    return MS

def convertisseur_km(mph):
    KH = mph * 1.609344
    return KH


convertisseur_en_ms = convertisseur_ms(vitesse)
print(f"{vitesse} miles/heure = {convertisseur_en_ms} mètres/seconde")
convertisseur_en_km = convertisseur_km(vitesse)
print(f"{vitesse} miles/heure = {convertisseur_en_km} kilomètres/heure")
```

**Écrire dans une cellule un programme qui demande à l'utilisateur de rentrer un entier n positif et qui affiche PAIR s'il est divisible par 2, IMPAIR sinon (penser au modulo).

```python
def entier_number (nombre_entier):
    if nombre_entier < 0:
        return "Votre nombre n'est pas positif"
    else:
        if nombre_entier % 2 == 0:
            return "PAIR"
        else:
            return "IMPAIR"

while True:
    n = int(input("Entrez votre nombre"))
    resultat = entier_number(n)
    if resultat == "Votre nombre n'est pas positif":
        print(resultat)
        break
    else:
        print(resultat)
```

```python
import cmath  # cmath permet de gérer les racines complexes

def resoudre_equation_second_degre(a, b, c):
    """
    Résout l'équation ax² + bx + c = 0.
    Retourne un tuple contenant les solutions (x1, x2) ou un message d'erreur.
    """
    # Validation des entrées
    try:
        a = float(a)
        b = float(b)
        c = float(c)
    except ValueError:
        return "Erreur : les coefficients doivent être des nombres."

    # Cas dégénéré : a = 0 → équation du premier degré ou constante
    if a == 0:
        if b == 0:
            return "Pas de solution" if c != 0 else "Infinité de solutions"
        else:
            return (-c / b,)  # Une seule solution

    # Calcul du discriminant
    delta = b**2 - 4*a*c

    # Calcul des solutions (réelles ou complexes)
    sqrt_delta = cmath.sqrt(delta)
    x1 = (-b - sqrt_delta) / (2*a)
    x2 = (-b + sqrt_delta) / (2*a)

    return (x1, x2)


# Exemple d'utilisation
if __name__ == "__main__":
    exemples = [
        (1, -3, 2),     # 2 solutions réelles distinctes
        (1, 2, 1),      # 1 solution réelle double
        (1, 0, 1),      # 2 solutions complexes
        (0, 2, -4),     # Équation du premier degré
        (0, 0, 5),      # Pas de solution
        (0, 0, 0)       # Infinité de solutions
    ]

    for a, b, c in exemples:
        print(f"Équation : {a}x² + {b}x + {c} = 0 → Solutions : {resoudre_equation_second_degre(a, b, c)}")
```
