**Liste de courses**

```python
import os
import json

FICHIER = "ma_liste.json"

#Affichage du menu
def afficher_menu():
    print("""
Choisissez une option:
\t1: Ajouter un élément
\t2: Enlever un élément
\t3: Afficher la liste
\t4: Vider la liste
\t5: Sauvegarder la liste
\t6: Quitter
""")

#Boucle pour que l'utilisateur fasse sa vie jusqu'à stopper le programme
while True:
    afficher_menu()
    choix = input("Votre choix : ")

    if choix == "1":
        element = input("Entrez l'élément à ajouter : ")
        liste.append(element)
        print(f"✅ '{element}' ajouté à la liste.")

    elif choix == "2":
        element = input("Entrez l'élément à enlever : ")
        try:
            liste.remove(element)
            print(f"🗑️ '{element}' retiré de la liste.")
        except ValueError:
            print("❌ Élément non trouvé dans la liste.")

    elif choix == "3":
        if liste:
            print("📋 Liste actuelle :")
            for i, item in enumerate(liste, 1):
                print(f"{i}. {item}")
        else:
            print("📭 La liste est vide.")

    elif choix == "4":
        liste.clear()
        print("🧹 Liste vidée.")

    elif choix == "5":
        try:
            with open(FICHIER, "w", encoding="utf-8") as f:
                json.dump(liste, f, ensure_ascii=False, indent=4)
            print(f"💾 Liste sauvegardée dans '{FICHIER}'.")
        except Exception as e:
            print("Erreur lors de la sauvegarde :", e)

    elif choix == "6":
        print("👋 Au revoir !")
        break

    else:
        print("❌ Option invalide. Veuillez choisir entre 1 et 6.")
```
