```python
def f(x):
    return 2*(x**3)+x-5

def poly(fonction, borneInf, borneSup, nbPas):
    pas=(borneSup-borneInf)/nbPas
    dico={}
    for i in range (nbPas+1):
        x=i*pas+borneInf
        y=fonction(x)
        dico[x]=y
    return dico

print(poly(f,0,5,5))
```

```python
def compterMots(phrase):
    dicoMots={} 
    mots=phrase.split(" ") 
    for mot in mots: 
        if mot not in dicoMots.keys(): 
            dicoMots[mot]=1 
        else :  
            dicoMots[mot]+=1    
    return dicoMots     

print(compterMots("Ceci est une petite phrase. Ou ceci est une grande phrase ?"))
```

```python
def minMaxMoy(liste):
    leMin=min(liste) 
    leMax=max(liste) 
    laMoy=sum(liste)/len(liste) 
    return (leMin,leMax,laMoy) 

resultat=minMaxMoy([10,18,14,20,12,16]) 
print(resultat)  
```
