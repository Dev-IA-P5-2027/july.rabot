**Écrire dans une cellule un programme qui demande à l'utilisateur de rentrer un rayon et la hauter, puis , qui calcule le volume d'un cône droit. (Rappel : Vcone = 1/3 * pi * r^2 * h) Bis : mettre ce programme dans une fonction volume_cone() et tester cette fonction**

```python
from math import pi

def volume_cone(rayon, hauteur):
    try:
        Volume = 1/3 * pi * rayon*rayon * hauteur
        return Volume
    except ValueError:
        return "Recommences tes valeurs sont pourris"

r = int(input("Entrez la valeur du rayon"))
h = int(input("Entrez la valeur de la hauteur"))
resultat_volume = volume_cone(r,h)
print(resultat_volume)
```

**Écrire une fonction qui convertit en mètres par seconde et en km/h une vitesse fournie en paramètres par l'utilisateur en miles/heure (1 mile = 1609 mètres). Ce résultat sera retourné par la fonction.**

```python
vitesse = int(input("Entrez votre vitesse en miles par heure : "))
print(f"Vous voulez convertir {vitesse} miles/heure")


def convertisseur_ms(mph):
    MS = mph * 0.44704
    return MS

def convertisseur_km(mph):
    KH = mph * 1.609344
    return KH


convertisseur_en_ms = convertisseur_ms(vitesse)
print(f"{vitesse} miles/heure = {convertisseur_en_ms} mètres/seconde")
convertisseur_en_km = convertisseur_km(vitesse)
print(f"{vitesse} miles/heure = {convertisseur_en_km} kilomètres/heure")
```

**Écrire dans une cellule un programme qui demande à l'utilisateur de rentrer un entier n positif et qui affiche PAIR s'il est divisible par 2, IMPAIR sinon (penser au modulo).
Bis : mettre ce programme dans une fonction pair() et tester cette fonction**

```python
def entier_number (nombre_entier):
    if nombre_entier < 0:
        return "Votre nombre n'est pas positif"
    else:
        if nombre_entier % 2 == 0:
            return "PAIR"
        else:
            return "IMPAIR"

while True:
    n = int(input("Entrez votre nombre"))
    resultat = entier_number(n)
    if resultat == "Votre nombre n'est pas positif":
        print(resultat)
        break
    else:
        print(resultat)
```

**Écrire un programme en python pour réviser ses tables de multiplication. Le programme tire au hasard deux entiers entre 0 et 10 et demande à l'utilisateur de donner leur produit.
Si l'utilisateur répond juste, il gagne un point. Si il répond faux, il perd un point.
On interrogera 5 fois l'utilisateur.
Le programme affiche à la fin le score de l'utilisateur.**

```python
import random

result = 0
score_user = 0

# Fonction pour multiplication
def multiplication(tirage1, tirage2):
    result = tirage1 * tirage2
    print(result)
    return result

# Fonction pour vérification du résultat du mec
def verification(user_nombre, correct_result):
    global score_user
    if user_nombre == correct_result:
        print("Votre résultat est Bon !")
        score_user += 1
    else:
        print("Votre résultat est Faux ...")
        score_user -= 1
    return score_user

# Boucle pour tirage
nombre_tirage = 0
while nombre_tirage < 5:
    tirage = [i for i in range(0, 11)]
    random.shuffle(tirage)
    premier_tirage = tirage[0]
    deuxieme_tirage = tirage[1]
    
    print(f"Le premier nombre tiré est : {premier_tirage}")
    print(f"Le deuxième nombre tiré est : {deuxieme_tirage}")

    # Demande au gars son résultat
    user_nombre = int(input("Entrez le résultat de la multiplication : "))
    correct_result = multiplication(premier_tirage, deuxieme_tirage)
    verification(user_nombre, correct_result)

    nombre_tirage += 1

print(f"Votre score final est : {score_user}")




```

**L'ordinateur tire un nombre entier au hasard entre 0 et 100.**
**L'utilisateur doit le trouver et pour cela propose des valeurs.**
**L'ordinateur indique pour chaque valeur proposée si la valeur est trop petite, trop grande ou s'il a trouvé.**
**a) Écrire un programme en Python pour jouer à ce jeu.**
**b) Modifier le programme pour qu'il s'arrête si l'utilisateur n'a pas trouvé au bout de 10 coups.**
**c) Mettre ce programme dans une fonction plus_ou_moins() qui prendra en paramètres :**
**le nombre min et max de la plage de tirage du nombre au hasard**
**le nombre de coups maximum**

```python
from random import randint

# Tire un nombre entre 1 et 100
nombre_aleatoire = randint(1, 100)

# Nombre d'essais
nombre_tirage = 0
max_tirages = 10

# Fonction nombre utilisateur conforme
def user_test(rep_user):
    if 0 <= rep_user <= 100:
        return True
    else:
        print("Votre nombre n'est pas entre 0 et 100, réessayez.")
        return False

# Fonction plus ou moins
def plus_ou_moins(user, aleatoire):
    if user > aleatoire:
        print("C’est plus petit.")
    elif user < aleatoire:
        print("C’est plus grand.")
    else:
        print("🎉 C’est gagné !")

# Boucle 
while nombre_tirage < max_tirages:
    try:
        nombre_user = int(input("Entrez un nombre entre 0 et 100 : "))
    except ValueError:
        print("Veuillez entrer un nombre valide.")
        continue
    if not user_test(nombre_user):
        continue
    nombre_tirage += 1
    plus_ou_moins(nombre_user, nombre_aleatoire)
    if nombre_user == nombre_aleatoire:
        print(f"Bravo ! Vous avez trouvé en {nombre_tirage} essai(s).")
        break
    print(f"Essai n°{nombre_tirage}/{max_tirages}")
else:
    print(f"Dommage 😢 Le nombre était {nombre_aleatoire}.")
```
