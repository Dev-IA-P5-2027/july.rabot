```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
```

```python
data = pd.read_csv("https://www.stat4decision.com/wine_data.csv", index_col = 0)
data["type"].value_counts()
```

```python
data.info()
```

```python
y = data["type"]
x = data.drop("type", axis = 1)
x_train, x_test, y_train, y_test  = train_test_split(x, 
                                                     y, 
                                                     test_size=0.25, 
                                                     random_state=42)
```

```python
modele_rf = RandomForestClassifier(
     n_estimators=100,
     criterion='gini',
     max_depth=None,
     min_samples_split=2,
     min_samples_leaf=1,
     min_weight_fraction_leaf=0.0,
     max_features='sqrt', #le 'auto' ne marchait pas 
     max_leaf_nodes=None,
     min_impurity_decrease=0.0,
     bootstrap=True,
     oob_score=False,
     n_jobs=None,
     random_state=None,
     verbose=0,
     warm_start=False,
     class_weight=None,
     ccp_alpha=0.0,
     max_samples=None,)
```

```python
modele_rf.fit(x_train, y_train)
```

```python
pd.DataFrame(modele_rf.feature_importances_,
              index = x_train.columns, 
              columns = ["importance"]).sort_values(
     "importance", 
     ascending = False)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix
print(f"Le pourcentage de bien classés est de : {accuracy_score(y_test, modele_rf.predict(x_test))*100} %")
```

```python
pd.DataFrame(confusion_matrix(y_test, modele_rf.predict(x_test)),
             index = ["blanc_données", "rouge_données"],
             columns = ["blanc_predit", "rouge_predit"])
```

*Corrigé*

```python
# 1. On s'assure de bien repartir de zéro pour X
# (Assurez-vous que 'df' contient bien vos données d'origine)
X = data.drop(['quality'], axis=1)  # On enlève la réponse
y = data['quality']                 # On garde la réponse à part

# 2. On refait le découpage (IMPORTANT : cela écrase vos anciens x_train/x_test pollués)
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# 3. On ré-entraîne le modèle
modele_rf.fit(X_train, y_train)

# 4. On affiche la NOUVELLE matrice de confusion
# (Avec le score réaliste qui devrait apparaître)
classes = modele_rf.classes_
pd.DataFrame(confusion_matrix(y_test, modele_rf.predict(X_test), labels=classes),
             index=classes, columns=classes)
```
