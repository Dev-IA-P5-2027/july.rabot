```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

```python
response = requests.get("http://localhost:8001").text
```

```python
soup = BeautifulSoup(response, 'lxml')
```

```python
soup.prettify
```

```python
table = soup.find('table', class_="visitors_table")
```

```python
# Vérifier que la page est bien récupérée
print(response)  # doit afficher du HTML, pas une chaîne vide

# Vérifier ce que BeautifulSoup voit
print(soup.prettify()[:500])  # affiche les 500 premiers caractères du HTML parsé

# Chercher tous les tableaux présents
print(soup.find_all('table'))  # si liste vide [] → aucun tableau trouvé
```

```python
result = {}
for row in table.find_all('tr'):
    data = row.find_all('td')
    for i in range(1, len(data), 2):
        result[data[i-1].text] = int(data[i].text)
```

```python
pd.DataFrame(result.items(), columns=['Mois', 'Visiteurs'])
```
