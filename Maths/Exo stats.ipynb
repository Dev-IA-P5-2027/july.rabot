```python
def calcul_impot(revenu):
    tranche1 = (11294, 0.0)  # (montant, taux, décote)
    tranche2 = (28797, 0.11)
    tranche3 = (82341, 0.30)
    tranche4 = (177106, 0.41)
    tranche5 = (200000, 0.45)
    
    decote_seuil = 1658
    decote_taux = 0.75
    
    impot = 0.0
    
    if revenu <= tranche1[0]:
        impot = tranche1[1] * revenu 
    elif revenu <= tranche2[0]:
        impot = tranche1[1] * tranche1[0] 
        impot += tranche2[1] * (revenu - tranche1[0])
    elif revenu <= tranche3[0]:
        impot = tranche1[1] * tranche1[0] 
        impot += tranche2[1] * (tranche2[0] - tranche1[0])
        impot += tranche3[1] * (revenu - tranche2[0])
    elif revenu <= tranche4[0]:
        impot = tranche1[1] * tranche1[0] 
        impot += tranche2[1] * (tranche2[0] - tranche1[0])
        impot += tranche3[1] * (tranche3[0] - tranche2[0])
        impot += tranche4[1] * (revenu - tranche3[0])
    else:
        impot = tranche1[1] * tranche1[0]
        impot += tranche2[1] * (tranche2[0] - tranche1[0])
        impot += tranche3[1] * (tranche3[0] - tranche2[0])
        impot += tranche4[1] * (tranche4[0] - tranche3[0])
        impot += tranche5[1] * (revenu - tranche4[0])
    
    if impot < decote_seuil:
        impot -= decote_taux * impot
    
    return impot

revenu = float(input("Entrez votre revenu annuel : "))
impot = calcul_impot(revenu)
print("L'impôt sur le revenu est de", round(impot, 2), "euros.")
```

```python
# Nombre de pièces
x = np.array([1, 2, 3, 4, 5, 6, 7])

# Nombre d'appartements (effectifs)
effectifs = np.array([48, 72, 96, 64, 39, 25, 3])

# Diagramme en bâtons
plt.bar(x, effectifs, color='pink', edgecolor='black')
plt.xlabel("Nombre de pièces")
plt.ylabel("Nombre d'appartements")
plt.title("Diagramme en bâtons des effectifs")
plt.show()

# Effectifs cumulés
effectifs_cumules = np.cumsum(effectifs)
print("Effectifs cumulés :", effectifs_cumules)
effectif_total = effectifs_cumules[-1]
print(f"L'effectif total est : {effectif_total}")

# --- 3. Fonction de distribution (répartition) ---
total = effectifs.sum()
F = effectifs_cumules / total
print("Fonction de répartition :", F)

# --- Graphique de la fonction de distribution ---
plt.figure(figsize=(5, 5))
plt.step(x, F, where='post', color='pink', linestyle='--')
plt.title("Fonction de distribution des effectifs")
plt.xlabel("Nombre de pièces")
plt.ylabel("F(x)")
plt.ylim(0, 1.05)
plt.grid(True, linestyle='-', alpha=0.4)
plt.show()

# Mode
mode = x[np.argmax(effectifs)]
print("Mode :", mode)

# Moyenne arithmétique
moyenne = np.sum(x * effectifs) / total
print("Moyenne :", moyenne)

# Étendue 
etendue = np.max(x) - np.min(x)
print("Étendue :", etendue)

# Variance 
variance = np.sum(effectifs * (x - moyenne)**2) / total
print("Variance :", variance)

# Écart-type 
ecart_type = np.sqrt(variance)
print("Écart-type :", ecart_type)
```

```python
# Classes : on prend les bornes inférieures et supérieures
classes = ["28-37", "38-47", "48-52", "53-57", "58-62", "63-72", "73-82"]
effectifs = np.array([3, 51, 74, 112, 92, 62, 6])

# Pour l'histogramme, on définit les bornes des classes
bornes = [28, 38, 48, 53, 58, 63, 73, 83]  # +1 pour inclure la dernière valeur
centres = np.array([(bornes[i] + bornes[i+1])/2 for i in range(len(effectifs))])

# Histogramme 
plt.figure(figsize=(8,4))
plt.hist(
    [ (bornes[i]+bornes[i+1])/2 for i in range(len(effectifs)) ],  # points centraux
    bins=bornes, 
    weights=effectifs, 
    edgecolor='black', 
    color='pink'
)
plt.xlabel("Masse de l'œuf (g)")
plt.ylabel("Nombre d'œufs")
plt.title("Histogramme des effectifs des œufs")
plt.grid(axis='y', linestyle='-', alpha=0.4)
plt.show()

#Effectifs cumulés
effectifs_cumules = np.cumsum(effectifs)
print("Effectifs cumulés :", effectifs_cumules)
effectif_total = effectifs_cumules[-1]
print(f"L'effectif total est : {effectif_total}")

#Fonction de distribution des effectifs 
total = effectifs.sum()
F = effectifs_cumules / total
print("Fonction de répartition :", F)

# Graphique de la fonction de distribution 
plt.figure(figsize=(5, 5))
plt.step(classes, F, where='post', color='pink', linestyle='--')
plt.title("Fonction de distribution des effectifs")
plt.xlabel("Masse de l'oeuf")
plt.ylabel("F(x)")
plt.ylim(0, 1.05)
plt.grid(True, linestyle='-', alpha=0.4)
plt.show()

# Classe modale
indice_modale = np.argmax(effectifs)
classe_modale = classes[indice_modale]
print(f"Classe modale : {classe_modale} g")


# Moyenne arithmétique
moyenne = np.sum(centres * effectifs) / total
print(f"Moyenne : {moyenne:.2f} g")

#Etendue
etendue = bornes[-1] - bornes[0]
print(f"Étendue : {etendue} g")

#Variance pondérée 
variance = np.sum(effectifs * (centres - moyenne)**2) / total
print(f"Variance : {variance:.2f} g²")

#Écart-type 
ecart_type = np.sqrt(variance)
print(f"Écart-type : {ecart_type:.2f} g")
```
