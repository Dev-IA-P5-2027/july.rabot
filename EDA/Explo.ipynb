**Faire un log pour mettre MedHouseVal en Normale pour la régression (limiter les erreurs)**

```python
# 1. On crée une nouvelle colonne avec le Log de la cible
# On utilise np.log1p (log de 1 + x) au cas où il y aurait des 0, c'est plus safe
df['MedHouseVal_Log'] = np.log1p(df['MedHouseVal'])

# 2. On compare les deux distributions
plt.figure(figsize=(12, 5))

# Graphique AVANT (Original)
plt.subplot(1, 2, 1)
sns.histplot(df['MedHouseVal'], kde=True, color="hotpink")
plt.title("Avant Log (Asymétrique)")

# Graphique APRÈS (Log)
plt.subplot(1, 2, 2)
sns.histplot(df['MedHouseVal_Log'], kde=True, color="deeppink")
plt.title("Après Log (Plus 'Normale')")

plt.tight_layout()
plt.show()
```

**Matrice de corrélation**

```python
# 1. On calcule les corrélations entre toutes les colonnes numériques
corr_matrix = df.corr()

# 2. On prépare la figure
plt.figure(figsize=(8, 6))

# 3. On dessine la heatmap
# annot=True affiche les chiffres dans les cases
# cmap='RdPu' pour le dégradé rose/violet
# fmt=".2f" pour n'avoir que 2 chiffres après la virgule
sns.heatmap(corr_matrix, annot=True, cmap='RdPu', fmt=".2f", linewidths=0.5)

plt.title("Qui est lié à quoi ? (Matrice de Corrélation)", fontsize=15)
plt.show()
```

**Analyse géographique : Créez un scatter plot avec Longitude en x, Latitude en y, et la couleur représentant MedHouseVal**

```python
# 2. On passe en MODE SOMBRE (Dark Mode)
plt.style.use('dark_background')
plt.figure(figsize=(12, 10))

# 3. Le Scatter Plot (Longitude en X, Latitude en Y)
# 'c' c'est pour la couleur (le prix)
# 'cmap' c'est le dégradé (flare ou RdPu pour rester dans le rose)
# 's' c'est la taille des points (10 pour pas que ce soit un gros pâté)
scatter = plt.scatter(df['Longitude'], df['Latitude'], 
                      c=df['MedHouseVal'], cmap='flare', 
                      s=10, alpha=0.6)

# On ajoute une barre de légende pour le prix
cbar = plt.colorbar(scatter)
cbar.set_label('Prix Médian (centaines de milliers de $)', color='white')

# On fignole le style
plt.title("Carte des prix de l'immobilier en Californie", color='hotpink', fontsize=15)
plt.xlabel("Longitude", color='hotpink')
plt.ylabel("Latitude", color='hotpink')

# On enlève les bordures inutiles pour le look
sns.despine(left=True, bottom=True)

plt.show()
```

```python

import geopandas as gpd
from shapely.geometry import Point


# 1. On prépare les données (comme avant)
from sklearn.datasets import fetch_california_housing
cal = fetch_california_housing()
df = pd.DataFrame(cal.data, columns=cal.feature_names)
df['Price'] = cal.target

# 2. LA MAGIE GEOPANDAS : On crée la colonne geometry
# On combine Longitude et Latitude pour faire des "Points"
geometry = [Point(xy) for xy in zip(df['Longitude'], df['Latitude'])]
gdf = gpd.GeoDataFrame(df, geometry=geometry)

# 3. Affichage style "Cyberpunk"
plt.style.use('dark_background')
fig, ax = plt.subplots(figsize=(12, 10))

# On dessine les points
# 'column' définit ce qui change la couleur
# 'markersize' pour la taille des points
gdf.plot(column='Price', ax=ax, cmap='RdPu', legend=True, 
         markersize=5, alpha=0.5,
         legend_kwds={'label': "Prix des maisons ($)"})

plt.title("Carte Géographique des Prix - Californie", color='hotpink', fontsize=15)
ax.set_axis_off() # On enlève les axes moches pour faire plus "carte"

plt.show()
```

```python
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point
import matplotlib.pyplot as plt
import contextily as ctx # La bibliothèque pour le fond de carte

# 1. Préparation des données
from sklearn.datasets import fetch_california_housing
cal = fetch_california_housing()
df = pd.DataFrame(cal.data, columns=cal.feature_names)
df['Price'] = cal.target

# 2. Création du GeoDataFrame et conversion de coordonnées
geometry = [Point(xy) for xy in zip(df['Longitude'], df['Latitude'])]
gdf = gpd.GeoDataFrame(df, geometry=geometry, crs="EPSG:4326")

# TRÈS IMPORTANT : On convertit en Web Mercator (mètres) pour la carte
gdf = gdf.to_crs(epsg=3857)

# 3. Le plot
plt.style.use('dark_background')
fig, ax = plt.subplots(figsize=(8, 10))

# On dessine les points (en rose 'flare' pour que ça pète)
gdf.plot(column='Price', ax=ax, cmap='flare', markersize=3, alpha=0.4, legend=True)

# 4. Ajout du fond de carte
# Source : CartoDB.DarkMatter (le plus stylé en mode sombre)
# Si tu veux le vrai OSM classique, utilise : ctx.providers.OpenStreetMap.Mapnik
ctx.add_basemap(ax, source=ctx.providers.CartoDB.DarkMatter)

plt.title("California Housing - Spatial Analysis", color='hotpink', fontsize=20)
ax.set_axis_off()

plt.show()
```
